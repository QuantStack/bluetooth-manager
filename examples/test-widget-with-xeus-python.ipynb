```python
import asyncio
from ipymovehub import MoveHubWidget,LedColor,Port, Sensor
from IPython.display import display
from ipymovehub.movehub import MoveHubLaneProxy

# the robot's LED supports the following colors
colors = [
    LedColor.pink,
    LedColor.purple,
    LedColor.blue,
    LedColor.lightblue,
    LedColor.cyan,
    LedColor.green,
    LedColor.yellow,
    LedColor.orange,
    LedColor.red,
]
```

# Connect the device 
Note that connecting the device should take 5-10 sec

```python
# number of concurrent "lanes"
n_lanes = 4
widget = MoveHubWidget(n_lanes=n_lanes)
display(widget)
```

```python
# Case 1 : Execute this cell if there is no MoveHub connected
widget.connect()
```

```python
# Case 2 : Execute this cell if there is/are one/some MoveHub(s) connected
# Choose one of them and paste its identifier
widget.identifier = "dcRIlxx9l8bQ3T6UVofW8Q=="
widget.connect(identifier=widget.identifier)
```

```python
display(widget)
```

# Change Color of LED

```python
lane_proxy = MoveHubLaneProxy (movehub = widget, lane = 0, identifier = widget.identifier)
for color in colors:
    print(color)
    #log(f"set color to {color.value}")
    await lane_proxy.set_led_async(color)
    await asyncio.sleep(2)
```

# Drive a polygon like path

```python
speed = 50
d_forward  = 200
d_rotate = 300

for i in range(4):
    await lane_proxy.set_led_async("red")
    await lane_proxy.motor_angle_multi_async(angle=d_forward, power_a=speed,power_b=speed)
    await lane_proxy.set_led_async("green")
    await lane_proxy.motor_angle_multi_async(angle=d_rotate, power_a=speed,power_b=-1.0*speed)
```
